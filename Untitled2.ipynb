```python
def ploting(var, num): 
    plt.subplot(2,2,num)
    sns.histplot(data=df[var], kde= True)
    

num_col = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR']

j = 1

for i in num_col: 
    ploting(i,j)
    j+=1
plt.tight_layout()
```

```python
col_mean_clean = df.loc[df['Cholesterol'] != 0, 'Cholesterol'].mean()
col_mean_clean = col_mean_clean.round(2)
```

```python
df['Cholesterol'] = df['Cholesterol'].replace(0, col_mean_clean)
```

```python
df['Cholesterol'].value_counts()
```

```python
col_mean_clean
```

```python
Re_mean_clean = df.loc[df['RestingBP'] != 0, 'RestingBP'].mean()
Re_mean_clean = Re_mean_clean.round(2)
```

```python
Re_mean_clean
```

```python
df['RestingBP'] = df['RestingBP'].replace(0, Re_mean_clean)
```

```python
def ploting(var, num): 
    plt.subplot(2,2,num)
    sns.histplot(data=df[var], kde= True)
    

num_col = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR']

j = 1

for i in num_col: 
    ploting(i,j)
    j+=1
plt.tight_layout()
```
