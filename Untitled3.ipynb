```python
# 1. Numeric → median  (using = instead of inplace=True)
num_cols = [
    'pack_capacity_kwh', 'battery_age_days', 'cycle_count',
    'avg_daily_distance_km', 'avg_charging_power_kw', 'fast_charge_ratio',
    'avg_depth_of_discharge_pct', 'avg_state_of_charge_pct',
    'charge_cycles_per_month', 'total_energy_throughput_kwh',
    'avg_ambient_temp_c', 'humidity_pct', 'altitude_m',
    'internal_resistance_mohm', 'voltage_avg_v', 'current_avg_a',
    'charge_time_avg_min', 'peak_temp_during_fast_charge_c',
    'service_count'
]
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# 2. Categorical → mode
cat_cols = ['chemistry', 'manufacturer', 'vehicle_model',
            'climate_zone', 'bms_firmware_version']
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# 3. recall_flag → fill with 0
df['recall_flag'] = df['recall_flag'].fillna(0)

# 4. Check — should all be 0 now
print(df.isnull().sum())
```

```python
# Define realistic ranges for each column
ranges = {
    'pack_capacity_kwh':              (10, 150),
    'battery_age_days':               (0, 5475),
    'cycle_count':                    (0, 5000),
    'avg_daily_distance_km':          (0, 800),
    'avg_charging_power_kw':          (0, 350),
    'fast_charge_ratio':              (0, 1),
    'avg_depth_of_discharge_pct':     (0, 100),
    'avg_state_of_charge_pct':        (0, 100),
    'charge_cycles_per_month':        (0, 200),
    'total_energy_throughput_kwh':    (0, 500000),
    'avg_ambient_temp_c':             (-40, 56.7),
    'humidity_pct':                   (0, 100),
    'altitude_m':                     (0, 5500),
    'internal_resistance_mohm':       (0, 500),
    'voltage_avg_v':                  (200, 900),
    'current_avg_a':                  (0, 1500),
    'charge_time_avg_min':            (5, 1440),
    'peak_temp_during_fast_charge_c': (10, 80),
    'service_count':                  (0, 30)
}

# Check each column
print("=" * 55)
print(f"{'Column':<35} {'Min':>6} {'Max':>10} {'Status'}")
print("=" * 55)

for col, (low, high) in ranges.items():
    actual_min = df[col].min()
    actual_max = df[col].max()
    
    # Flag if outside realistic range
    flag = ""
    if actual_min < low:
        flag += "❌ MIN too low! "
    if actual_max > high:
        flag += "❌ MAX too high!"
    if flag == "":
        flag = "✅ OK"
    
    print(f"{col:<35} {actual_min:>6.1f} {actual_max:>10.1f}   {flag}")

print("=" * 55)
```
