# Data Cleaning and preprocessing 

```python
df_c = df.copy()
```

```python
df_c
```

```python
df_c.drop_duplicates(inplace= True)
```

```python
df_c
```

```python
df_c.isnull().sum()
```

```python
df_c.dtypes
```

```python
df_c['sex'].value_counts()
```

```python
df_c['sex'] = df_c['sex'].map({'male': 0, 'female': 1})
```

```python
df_c
```

```python
df_c['smoker'].value_counts() 
```

```python
df_c['smoker'] = df_c['smoker'].map({'yes' : 1, 'no': 0})
```

```python
df_c
```

```python
df_c['region'].value_counts()
```

```python
df_c.rename(columns={'sex' : 'is_female', 'smoker': 'is_smoker'}, inplace= True)
```

```python
df_c
```

```python
df_c = pd.get_dummies(df_c,columns=['region'])
```

```python
df_c
```

```python
df_c.astype(int)
```

```python
df_c = df_c.astype(int)
```
