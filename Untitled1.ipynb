# Feature Engineering and Extraction 

```python
df_c['bmi_category'] = pd.cut(
    df_c['bmi'],                          # the column to bin
    bins=[0, 18.5, 24.9, 29.9, float('inf')],   # the cut points
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']  # names for each bin
)
```

```python
df_c
```

```python
df_c = pd.get_dummies(df_c, columns=['bmi_category']) 
```

```python
df_c
```

```python
df_c = df_c.astype(int)
```

```python
df_c
```

```python
df_c.columns
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
col = ['age', 'bmi', 'children']
```

```python
scal = StandardScaler()
```

```python
df_c[col] = scal.fit_transform(df_c[col])
```

```python
df_c
```

```python
df_c['age'].min()
```

```python
# Correlation of all numeric columns with the target (charges)
correlation = df_c.select_dtypes(include=['int64', 'float64']).corr(method='pearson')['charges'].sort_values(ascending=False)
print(correlation)
```

```python
df_c['charges_category'] = pd.qcut(
    df_c['charges'],
    q=4,
    labels=['Low', 'Medium', 'High', 'Very High']
)
```

```python
categorical_cols = [
    'is_smoker', 'is_female',
    'region_northeast', 'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Underweight', 'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]
```

```python
from scipy.stats import chi2_contingency

results = []

for col in categorical_cols:
    contingency_table = pd.crosstab(df_c[col], df_c['charges_category'])
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    
    results.append({
        'Feature': col,
        'Chi2 Statistic': round(chi2, 4),
        'p-value': round(p, 4),
        'Decision': 'Keep' if p < 0.05 else 'Drop'
    })

chi_square_results = pd.DataFrame(results).sort_values(by='Chi2 Statistic', ascending=False).reset_index(drop=True)
print(chi_square_results)
```
